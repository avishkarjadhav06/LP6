```
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
from sklearn.preprocessing import MinMaxScaler

# Create sample stock price data
data = {
    'Price': [100, 102, 101, 105, 107,
              110, 108, 111, 115, 117,
              120, 119, 123, 125, 128]
}

df = pd.DataFrame(data)

# Normalize data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)

# Prepare dataset (time-series window of 3)
X = []
y = []

for i in range(3, len(scaled_data)):
    X.append(scaled_data[i-3:i, 0])
    y.append(scaled_data[i, 0])

X = np.array(X)
y = np.array(y)

# Reshape for RNN (samples, timesteps, features)
X = np.reshape(X, (X.shape[0], X.shape[1], 1))

# Build RNN model
model = Sequential()

model.add(SimpleRNN(50, activation='relu', input_shape=(X.shape[1], 1)))
model.add(Dense(1))

# Compile model
model.compile(optimizer='adam', loss='mean_squared_error')

# Train model
model.fit(X, y, epochs=100, verbose=1)

# Prediction
prediction = model.predict(X)

# Convert back to original scale
prediction = scaler.inverse_transform(prediction)

print("Predicted Stock Prices:")
print(prediction)
```
